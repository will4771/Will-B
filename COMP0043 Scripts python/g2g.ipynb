```python
import numpy as np
import matplotlib.pyplot as plt
```

Guassian to Gussian Fourier Pair

```python
# Parameters
N = 1024
Dx = 0.1
x = Dx * np.arange(-N / 2, N / 2)  # Spatial domain
L = N * Dx
Dxi = 2 * np.pi / L  # Nyquist relation: Dx * Dxi = 2 * pi / N
xi = Dxi * np.arange(-N / 2, N / 2)  # Frequency domain
W = N * Dxi  # = 2 * pi / Dx
```

```python
# Gaussian parameters
a = 1
fa = np.sqrt(a / np.pi) * np.exp(-a * x**2)  # Gaussian in x space
Fa = np.exp(-xi**2 / (4 * a))  # Gaussian in xi space
```

```python
Fn = np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(fa))) * L 
# 1) np.fft.ifftshift(fa) : Reorders the data so that the origin (zero frequency) is at the start, matching FFT conventions.
# 2) np.fft.ifft(...): Computes the inverse Fourier transform of the reordered data, resulting in the frequency domain representation.
# 3) np.fft.fftshift(...): Reorders the result back to have zero frequency at the center, suitable for plotting and analysis.
# 4) * L: Multiplies by the spatial domain length L, accounting for normalization to match continuous Fourier transform conventions.



fn = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(Fa))) / L
Fn1 = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(fa))) * Dx
fn1 = np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(Fa))) / Dx
```

```python

plt.plot(x, np.real(fn), 'r', label='Re(fn)')
plt.plot(x, np.imag(fn), 'g', label='Im(fn)')
plt.plot(x, fa, 'p', label='fa')
plt.axis([-5, 5, 0, 1])
plt.xlabel('x')
plt.ylabel('f')
plt.legend()
plt.title('Comparison of Fourier Transforms in Spatial Domain')
```

```python
plt.plot(xi, np.real(Fn), 'r', label='Re(Fn)')
plt.plot(xi, np.imag(Fn), 'm', label='Im(Fn)')
plt.plot(xi, Fa, 'p:', label='Fa')
plt.axis([-10, 10, 0, 1])
plt.xlabel(r'$\xi$')
plt.ylabel('F')
plt.legend()
plt.title('Comparison of Fourier Transforms in Frequency Domain')
```

```python
plt.plot(x, np.real(fn1), 'r', label='Re(fn1)')
plt.plot(x, np.imag(fn1), 'g', label='Im(fn1)')
plt.plot(x, fa, 'p:', label='fa')
plt.axis([-5, 5, 0, 1])
plt.xlabel('x')
plt.ylabel('f')
plt.legend()
plt.title('Alternate Normalization in Spatial Domain')
```

```python
plt.plot(xi, np.real(Fn1), 'b', label='Re(Fn1)')
plt.plot(xi, np.imag(Fn1), 'm', label='Im(Fn1)')
plt.plot(xi, Fa, 'p:', label='Fa')
plt.axis([-10, 10, 0, 1])
plt.xlabel(r'$\xi$')
plt.ylabel('F')
plt.legend()
plt.title('Alternate Normalization in Frequency Domain')
```
