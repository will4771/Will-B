```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon
import random
```

```python
mu = 2/3 # meam
nsamples = 10**6 # number of random samples
a = 0 # left truncation
b = 5 # right truncation 
n = 100 # grids

x = np.linspace(a, b, n)
```

```python
# Computing the CDF and PDF

f = 1/mu * np.exp(-x/mu)
F = 1 - np.exp(-x/mu)

sf = expon.pdf(x, scale = mu) # have to use scale 
sF = expon.cdf(x, scale = mu)
```

```python
# Plotting pdf

plt.plot(x, f, 'r', label = 'theoretical pdf' )
plt.plot(x, sf, 'g--', label = 'scipy pdf')
plt.xlabel('x')
plt.ylabel('pdf')
plt.title('Pdf of exponential distribution wuth mu = 2/3')
```

```python
# Plotting CDF

plt.plot(x, F, 'r', label = 'theoretical cdf' )
plt.plot(x, sF, 'g--', label = 'scipy cdf')
plt.xlabel('x')
plt.ylabel('cdf')
plt.title('cdf of exponential distribution wuth mu = 2/3')
```

```python
# Sampling an exponetional distribtuion using X = -mu * log(U)

U = np.random.uniform(0,1, nsamples)
X = - mu * np.log(U)
```

```python
# plotting histogram 
plt.hist(X, bins = 1000, density=True, alpha=0.6, color='b', label='Truncated Exponential')
plt.plot(x, f, 'r--', label = 'theortical pdf')
plt.xlabel('x')
plt.ylabel('pdf')
plt.title('sampled exponetial with mu = 2/3')
plt.legend()
plt.show()

```
