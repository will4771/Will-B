```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import lognorm
```

```python
mu = 0.2 # mean
sigma = 0.1 # sigma
nsample = 10**6 # number of samples
a = 0 # left truncation
b = 2 # right truncation
n = 101 # number of samples

x = np.linspace(a, b, n)
```

```python
# computing pdf and cdf

f = (1 / (x * sigma * np.sqrt(2 * np.pi))) * np.exp(-((np.log(x) - mu) / sigma) ** 2 / 2)

sf = lognorm.pdf(x, sigma, scale=np.exp(mu))
sF = lognorm.cdf(x, sigma, scale=np.exp(mu))
```

```python
# plotting pdf of log normal

plt.plot(x, f, 'b', label = 'theoritcal pdf')
plt.plot(x, sf, 'r--', label = 'scipy pdf')
plt.xlabel('x')
plt.ylabel('Density')
plt.legend()
plt.title('Log-normal Distribution: Theoretical vs SciPy')
```

```python
# log normal CDF 

plt.plot(x ,sF, label= 'scippy cdf')
plt.xlabel('x')
plt.ylabel('CDF')
plt.legend()
plt.title('Log normal CDF')
```

```python
# Sampled Log normal. 

# first we sample a normal 

from scipy.stats import norm

U = np.random.uniform(0, 1, nsample)

X = mu + sigma * norm.ppf(U)

Y = np.exp(X)
```

```python
plt.hist(Y, bins=100, density=True, alpha=0.6, color='g')
plt.plot(x, sf, 'r--', label = 'scipy pdf')

plt.show()
```
