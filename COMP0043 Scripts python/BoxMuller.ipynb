```python
import numpy as np
import matplotlib.pyplot as plt
import random
from scipy.stats import norm
```

```python
nsamples = 10**7 # number of samples
a = -4 # left truncation
b = 4 # right truncation
ngrid = 100 # number of partiti
mu = 0 # mean of std norm
sig = 1 #std of std norm
xgrid = np.linspace(a, b, ngrid) # x grid
```

```python
U1 = np.random.uniform(0, 1, nsamples)
U2 = np.random.uniform(0, 1, nsamples)

Y1 = np.sqrt(-2 * np.log(U1)) * np.cos(2*np.pi*U2)
```

```python
plt.hist(Y1, bins = 1000, density=True, label = 'box-muller')
plt.plot(xgrid, norm.pdf(xgrid, mu, sig), 'r--', label = 'analtical pdf')
plt.xlabel('x')
plt.ylabel('pdf')
plt.title('pdf of std normal with Box Muller')
plt.show()
```
