```python
import numpy as np
import matplotlib.pyplot as plt 
from mpl_toolkits.mplot3d import Axes3D
```

```python
# Linear Congruential generator
a = 1597
b = 51749
M = 244944

def rand(n,N):
    res = np.zeros(n)
    res[0] = N

    for i in range(1,n):
        res[i] = ((a * res[i - 1] + b) % M)
        
    return [x / M for x in res]

```

```python
# histrogram of our LCG

nsamples = 10000
N = 9

plt.hist(rand(nsamples,N), bins = 100, density=True )
plt.xlabel("x")
plt.ylabel('frequencey')
plt.title('linear congruential generator')
plt.show()
```

```python
plt.scatter(rand(nsamples, N), rand(nsamples, N - 1))
plt.xlabel('U_N')
plt.ylabel('U_N-1')
```
