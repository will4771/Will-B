```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
N = 1024
Dx = 0.1
x = Dx * np.arange(-N/2, N/2)
L = N * Dx

Dxi = 2 * np.pi / L
xi = Dxi * np.arange(-N/2, N/2)
W = N * Dxi
```

```python
# Approximate Dirac Delta using a narrow Gaussian function
a = 100  # Width parameter for the Gaussian (larger 'a' makes it more delta-like)
fa = np.exp(-a * x**2)  # Gaussian function

# Normalize so that the integral is 1 (for Dirac Delta property)
fa /= np.sum(fa) * Dx  # Normalizing factor

Fa = np.ones(len(x))
```

```python
Fn = np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(fa))) * L
fn = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(Fa))) / L

Fn1 = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(fa))) * Dx
fn1 = np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(Fa))) / Dx
```

```python
plt.plot(x, np.real(fn), 'r', label = 'Re(fn)')
plt.plot(x, np.imag(fn), 'g', label = 'Im(fn)')
plt.plot(x, fa, 'b--', label = fa)
plt.axis([-1, 1, 0, 12])
plt.xlabel('x')
plt.ylabel('f')
plt.legend()
plt.title('Comparison of Fourier Transforms in Spatial Domain')
plt.show()
```

```python
plt.plot(xi, np.real(Fn), 'r', label = 'Re(Fn)')
plt.plot(xi, np.imag(Fn), 'g', label = 'Im(Fn)')
plt.plot(xi, Fa, 'b--', label = fa)
plt.axis([-20, 20, 0, 2])
plt.xlabel(r'$\nu$')
plt.ylabel('F')
plt.legend()
plt.title('Fourier Transform (Frequency Domain)')
plt.show()
```

```python
plt.plot(x, np.real(fn1), 'r', label = 'Re(fn1)')
plt.plot(x, np.imag(fn1), 'g', label = 'Im(fn1)')
plt.plot(x, fa, 'b--', label = fa)
plt.axis([-1, 1, 0, 12])
plt.xlabel('x')
plt.ylabel('f')
plt.legend()
plt.title('Inverse Fourier Transform with Alternate Scaling ')
plt.show()
```

```python
plt.plot(xi, np.real(Fn1), 'r', label = 'Re(Fn)')
plt.plot(xi, np.imag(Fn1), 'g', label = 'Im(Fn)')
plt.plot(xi, Fa, 'b--', label = fa)
plt.axis([-20, 20, 0, 2])
plt.xlabel(r'$\nu$')
plt.ylabel('F')
plt.legend()
plt.title('Fourier Transform with Alternate Scaling (Frequency Domain)')
plt.show()
```
