```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Monte Carlo Simulation of arithmetic brownian motion
# dX(t) = mu * dt + sigma * dW(t)
```

```python
# Parameters and grid

npaths = 20000 # number of paths
T = 1 # Time horizon
nsteps = 200 # number of time steps
dt = T / nsteps # time steps
t = np.arange(0, T + dt, dt) # observation times
mu = 0.12 # model parameters
sigma = 0.4 # model parameters

```

```python
# Compute the increments with Euler-Maruyama
dX = mu * dt + sigma * np.random.randn(npaths, nsteps) * np.sqrt(dt)

# Accumulate the increments
X = np.hstack((np.zeros((npaths, 1)), np.cumsum(dX, axis=1)))
```

```python

# Expected path and sample mean
EX = mu * t  # Expected path
mean_X = np.mean(X, axis=0)
```

```python
# Plot the results
plt.figure(figsize=(10, 6))
plt.plot(t, EX, 'k', label='Expected path')
plt.plot(t, mean_X, 'r--', label='Mean path')
plt.plot(t, X[::1000].T, alpha=0.5)  # Plot a subset of paths for clarity
plt.legend()
plt.xlabel('t')
plt.ylabel('X')
plt.ylim([-1, 1])
plt.title(r'Arithmetic Brownian motion $dX(t) = \mu dt + \sigma dW(t)$')
plt.show()
```
