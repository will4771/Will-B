```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# Parameters

N = 1024
Dx = 0.1
x = Dx * np.arange(-N/2, N/2) # spatial domain
L = N * Dx

Dxi = 2 * np.pi / L # from Nyquist relation
xi = Dxi * np.arange(-N/2, N/2) 
W = N * Dxi
```

```python
# Distributions 

a = 1

fa = a/2 * np.exp(-a *np.abs(x)) # Laplace
Fa = a**2 / (a**2 + xi**2) # Lorentzian or Cauchy
```

```python
Fn =  np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(fa))) * L # in spatial domain
fn =  np.fft.fftshift(np.fft.fft(np.fft.ifftshift(Fa))) / L # in spatial domain

Fn1 =  np.fft.fftshift(np.fft.fft(np.fft.ifftshift(fa))) * Dx # we using foward from fourer space here
fn1 =  np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(Fa))) / Dx # using inverse from x space

```

```python
plt.plot(x, np.real(fn), 'r', label = 'Re(fn)')
plt.plot(x, np.imag(fn), 'g', label = 'Im(fn)')
plt.plot(x, fa, 'b--', label='fa')
plt.axis([-10, 10, 0, 1])
plt.xlabel('x')
plt.ylabel('f')
plt.legend()
plt.title('Comparison of Fourier Transforms in Spatial Domain')
```

```python
plt.plot(xi, np.real(Fn), 'r', label = 'Re(Fn)')
plt.plot(xi, np.imag(Fn), 'g', label = 'Im(Fn)')
plt.plot(xi, Fa, 'b--', label='Fa')
plt.axis([-10 / np.pi, 10 / np.pi, 0, 1])
plt.xlabel(r'$\nu$')
plt.ylabel('F')
plt.legend()
plt.title('Fourier Transform (Frequency Domain)')
```

```python
plt.plot(x, np.real(fn1), 'r', label = 'Re(fn)')
plt.plot(x, np.imag(fn1), 'g', label = 'Im(fn)')
plt.plot(x, fa, 'b--', label='fa')
plt.axis([-10, 10, 0, 1])
plt.axis([-10, 10, 0, 1])
plt.xlabel('x')
plt.ylabel('f')
plt.legend()
plt.title('Inverse Fourier Transform with Alternate Scaling ')
```

```python
plt.plot(xi, np.real(Fn1), 'b', label='Re(Fn1)')
plt.plot(xi, np.imag(Fn1), 'm', label='Im(Fn1)')
plt.plot(xi, Fa, 'c:', label='Fa')
plt.axis([-10 / np.pi, 10 / np.pi, 0, 1])
plt.xlabel(r'$\nu$')
plt.ylabel('F')
plt.legend()
plt.title('Fourier Transform with Alternate Scaling (Frequency Domain)')
```
