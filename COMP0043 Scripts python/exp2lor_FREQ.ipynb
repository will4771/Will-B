```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# Parameters
# Parameters
N = 1024
Dx = 0.1
x = Dx * np.arange(-N / 2, N / 2)  # Time domain
Lx = N * Dx

Dnu = 1 / Lx  # Nyquist relation: Dx * Dnu = 1 / N
nu = Dnu * np.arange(-N / 2, N / 2)  # Frequency domain
Nu = N * Dnu  # = 1 / Dt
```
