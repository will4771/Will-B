```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
nsamples = 10000 # number of samples
a = 0 # left truncation
b = 10 # right truncation
ngrid = 10000 # grid partitions
xgrid = np.linspace(a,b,ngrid) # grid
```

```python
U = np.random.uniform(-np.pi/2,np.pi/2,nsamples)

X = np.cos(U)

f = 2 / (np.pi * np.sqrt(1 - xgrid**2))
```

```python
plt.hist(X, bins = 100, density=True, label = 'sampled result')
plt.plot(xgrid, f, 'r', label ='therotical result ' )
plt.xlabel('x')
plt.ylabel('pdf')
plt.title('pdf of X = cosU, U ~ (-pi/2, pi.2)')
plt.show()
```
