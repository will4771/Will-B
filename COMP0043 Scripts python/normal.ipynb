```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
mu = 2 # mean
sigma = 1 # standard deviation
a = -5 # left truncation
b = 5 # right truncation
ngrid = 200 # number of grid intervals
nsample = 10**6 # number of random samples

x = np.linspace(a, b, ngrid) # grid
```

```python
f1 = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp((-((x - mu) ** 2) / (2 * sigma ** 2)))

f = norm.pdf(x, mu, sigma)
F = norm.cdf(x, mu, sigma)
```

```python
# plotting normal distribution

plt.plot(x, f1, 'g', label = 'Theortical pdf')
plt.plot(x, f, 'r--', label = 'Scipy pdf')
plt.plot(x, F, 'b', label = 'CDF')
plt.xlabel('x')
plt.title('pdf and cdf of normal dis')
```

```python
# Sampled normal distribution 

U = np.random.uniform(0, 1, nsample)
X1 = mu + sigma * norm.ppf(U)

plt.hist(X1, bins = 1000, density=True)
plt.title('method 1')
plt.show()
```
