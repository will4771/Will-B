```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Linear Congruential generator
a = 1366
b = 150889
M = 714025

def rand(n,N):
    res = np.zeros(n)
    res[0] = N

    for i in range(1,n):
        res[i] = ((a * res[i - 1] + b) % M)
        
    return [x / M for x in res]

```

```python
def Fib_gen(n,N): # only for n > 17
    res = rand(17,N)

    for i in range(17,n):
        U = res[i - 17] - res[i - 5]
        if U < 0:
            U +=1
        res.append(U)
    
    return res
```

```python
nsample = 1000
N = 109


UN1 = Fib_gen(nsample,N)
UN2 = Fib_gen(nsample,N -1)

plt.scatter(UN1,UN2)
```
